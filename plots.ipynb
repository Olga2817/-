```python
import matplotlib.pyplot as plt
import numpy as np

data = np.loadtxt('simulation_results.txt')

rounds = data[:, 0]
balances = data[:, 1]

plt.figure(figsize=(10, 5))
plt.plot(rounds, balances, marker='o', color='purple', linewidth=2, label='Баланс из C++')

plt.title('График симуляции казино по данным из C++', fontsize=14, fontweight='bold')
plt.xlabel('Номер раунда', fontsize=12)
plt.ylabel('Баланс (в фишках)', fontsize=12)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
